```python
## Answering common questions
##Total revenue?
total_revenue = order_details['UnitPrice'].sum()
print("Total Revenue: " + str(total_revenue.round(2)))

##Total profit?
total_profit = order_details['UnitPrice'].sum() - order_details['UnitCost'].sum()
print("Total Profit: " + str(total_profit.round(2)))

##Total number of orders?
total_orders = orders['OrderID'].nunique()
print("Total Orders: " + str(total_orders))

##Number of products sold?
total_products_sold = order_details['ProductID'].nunique()
print("Total Products Sold: " + str(total_products_sold))
```

```python
##Product and category questions
##Product with the most revenue?
product_w_most_revenue = (
    order_details.groupby('ProductID', as_index=False)['UnitPrice']
    .sum()
    .rename(columns={'UnitPrice': 'Revenue'})
    .merge(products[['ProductID', 'ProductName']], on='ProductID', how='left')
    .sort_values('Revenue', ascending=False)
    .head()
)
product_w_most_revenue
```

```python
##Product with most profit?
product_w_most_profit = (
    order_details.assign(
        Profit=order_details['UnitPrice'] - order_details['UnitCost']
    )
    .groupby('ProductID', as_index=False)['Profit']
    .sum()
    .merge(
        products[['ProductID', 'ProductName']],
        on='ProductID',
        how='left'
    )
    .sort_values('Profit', ascending=False)
)

product_w_most_profit.head()
```

```python
##Are the highest selling products the most profitable?
highest_selling = order_details.groupby('ProductID')['Quantity'].sum().reset_index(name='TotalQuantitySold')
is_most_profitable = order_details.groupby('ProductID').apply(lambda x: (x['UnitPrice'] - x['UnitCost']).sum()).reset_index(name='TotalProfit')
comparison_result = highest_selling.merge(
    is_most_profitable,
    on='ProductID',
    how='left'
)
comparison_result.head()
```

```python
##Percentage of profit margin by category
category_margin_pct = pd.read_sql_query("""
SELECT cat.CategoryName,
    (SUM(od.Quantity * od.UnitPrice * (1 - od.DiscountRate)) - SUM(od.Quantity * od.UnitCost))
        / SUM(od.Quantity * od.UnitPrice * (1 - od.DiscountRate)) * 100 AS MarginPct -- profit as a % of revenue
FROM categories AS cat
INNER JOIN products AS p
    ON cat.CategoryID = p.CategoryID
INNER JOIN order_details AS od
    ON p.ProductID = od.ProductID
GROUP BY cat.CategoryName
ORDER BY MarginPct DESC;
"""
, connection)
```

```python
##Percentage margin by category 
sorted_margin = category_margin_pct.sort_values('MarginPct', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=sorted_margin,
    x='CategoryName',
    y='MarginPct',
    color='green'
)

plt.xlabel('Product Category')
plt.ylabel('Margin (% of Revenue)')
plt.title('Profit Margin (%) by Category')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
