```python
## Answering common questions
##Total revenue?
total_revenue = order_details['UnitPrice'].sum()
print("Total Revenue: " + str(total_revenue.round(2)))

##Total profit?
total_profit = order_details['UnitPrice'].sum() - order_details['UnitCost'].sum()
print("Total Profit: " + str(total_profit.round(2)))

##Total number of orders?
total_orders = orders['OrderID'].nunique()
print("Total Orders: " + str(total_orders))

##Number of products sold?
total_products_sold = order_details['ProductID'].nunique()
print("Total Products Sold: " + str(total_products_sold))
```

```python
##Product and category questions
##Product with the most revenue?
product_w_most_revenue = order_details.groupby('ProductID')['UnitPrice'].sum().idxmax()
print("Product with Most Revenue: " + str(product_w_most_revenue))

##Product with most profit?
product_w_most_profit = order_details.groupby('ProductID').apply(lambda x: (x['UnitPrice'] - x['UnitCost']).sum()).idxmax()
print("Product with Most Profit: " + str(product_w_most_profit))

##Are the highest selling products the most profitable?
highest_selling = order_details.groupby('ProductID')['Quantity'].sum().idxmax()
is_most_profitable = product_w_most_profit == highest_selling
print("Is the highest selling product also the most profitable? " + str(is_most_profitable))
```
